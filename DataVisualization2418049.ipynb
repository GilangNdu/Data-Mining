### **2. Import Library Dan Membaca Dataset**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca dataset yang sudah dibersihkan
df = pd.read_excel("Dataset_Mahasiswa_Clean.xlsx")

# Mengatur tampilan grafik
sns.set_theme(style="whitegrid")

# Memastikan data berhasil dibacas
print("Ukuran dataset:", df.shape)
display(df.head())
```

**3. Histogram Distribusi IPK**

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="IPK",
    bins=15,
    kde=True
)

plt.title("Distribusi IPK Mahasiswa")
plt.xlabel("Indeks Prestasi Kumulatif (IPK)")
plt.ylabel("Jumlah Mahasiswa")
plt.tight_layout()
plt.show()
```

**4. Distribusi Kehadiran Mahasiswa**

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Kehadiran (%)",
    bins=15,
    kde=True
)

plt.title("Distribusi Kehadiran Mahasiswa")
plt.xlabel("Kehadiran (%)")
plt.ylabel("Jumlah Mahasiswa")
plt.tight_layout()
plt.show()
```

**5. Scatter Plot Kehadiran Terhadap IPK**

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Kehadiran (%)",
    y="IPK",
    scatter_kws={"alpha": 0.6},
    line_kws={"color": "red"}
)

plt.title("Hubungan Kehadiran dengan IPK Mahasiswa")
plt.xlabel("Kehadiran (%)")
plt.ylabel("IPK")
plt.tight_layout()
plt.show()
```

**6. Visualisasi Aktivitas Pembelajaran Terhadap IPK**

```python
aktivitas = [
    "Partisipasi Diskusi (skor)",
    "Nilai Tugas (rata-rata)",
    "Aktivitas E-Learning (skor)"
]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, kolom in zip(axes, aktivitas):
    sns.regplot(
        data=df,
        x=kolom,
        y="IPK",
        ax=ax,
        scatter_kws={"alpha": 0.5},
        line_kws={"color": "red"}
    )

    ax.set_title(f"{kolom} vs IPK")
    ax.set_ylabel("IPK")

plt.suptitle(
    "Hubungan Aktivitas Pembelajaran dengan IPK",
    fontsize=15
)
plt.tight_layout()
plt.show()
```

**7. Heatmap Korelasi Antarvariabel Numerik**

```python
kolom_numerik = [
    "Umur",
    "Kehadiran (%)",
    "Partisipasi Diskusi (skor)",
    "Nilai Tugas (rata-rata)",
    "Aktivitas E-Learning (skor)",
    "IPK"
]

korelasi = df[kolom_numerik].corr(numeric_only=True)

plt.figure(figsize=(10, 7))

sns.heatmap(
    korelasi,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Heatmap Korelasi Variabel Numerik")
plt.tight_layout()
plt.show()
```

**9. Catat Pola Dan Identifikasi Variabel**

```python
import pandas as pd
from scipy.stats import pearsonr
import os

# Folder untuk menyimpan hasil analisis
folder = "/content/Hasil_Data_Visualization"
os.makedirs(folder, exist_ok=True)

# Variabel aktivitas pembelajaran (disesuaikan dengan nama kolom asli)
variabel = [
    "Kehadiran (%)",
    "Partisipasi Diskusi (skor)",
    "Nilai Tugas (rata-rata)",
    "Aktivitas E-Learning (skor)"
]

# Menghitung korelasi setiap variabel dengan IPK
hasil = []

for kolom in variabel:
    data = df[[kolom, "IPK"]].apply(
        pd.to_numeric, errors="coerce"
    ).dropna()

    if len(data) >= 3 and data.nunique().min() > 1:
        r, p = pearsonr(data[kolom], data["IPK"])

        hasil.append({
            "Variabel": kolom,
            "Jumlah Data": len(data),
            "Korelasi dengan IPK": round(r, 3),
            "Nilai p": round(p, 4)
        })

hasil_df = pd.DataFrame(hasil)

# Urutkan berdasarkan kekuatan korelasi absolut
hasil_df["Kekuatan Korelasi"] = (
    hasil_df["Korelasi dengan IPK"].abs()
)

hasil_df = hasil_df.sort_values(
    "Kekuatan Korelasi", ascending=False
)

# Tampilkan hasil
display(hasil_df)

# Simpan tabel identifikasi
hasil_df.to_csv(
    f"{folder}/identifikasi_variabel.csv",
    index=False
)

# Catat ringkasan otomatis
with open(
    f"{folder}/catatan_pola_visualisasi.txt",
    "w",
    encoding="utf-8"
) as f:
    f.write("CATATAN HASIL DATA VISUALIZATION\n\n")

    for _, baris in hasil_df.iterrows():
        r = baris["Korelasi dengan IPK"]

        if r > 0:
            arah = "positif"
        elif r < 0:
            arah = "negatif"
        else:
            arah = "tidak menunjukkan korelasi linear"

        f.write(
            f"{baris['Variabel']}: korelasi {arah}, "
            f"r = {r:.3f}, p = {baris['Nilai p']:.4f}\n"
        )

    if not hasil_df.empty:
        terkuat = hasil_df.iloc[0]
        f.write(
            "\nVariabel dengan korelasi absolut terbesar: "
            f"{terkuat['Variabel']} "
            f"(r = {terkuat['Korelasi dengan IPK']:.3f}).\n"
        )

print("Hasil identifikasi dan catatan berhasil disimpan.")
```
